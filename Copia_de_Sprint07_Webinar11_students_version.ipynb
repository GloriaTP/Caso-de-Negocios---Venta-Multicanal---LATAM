## Ejercicio 3 · Crear una copia limpia \(15 min\)

Regla de trabajo: no modificamos el dataset original directamente. Creamos `df_limpio` para preservar la trazabilidad.

Aplicaremos cuatro decisiones:

1. Convertir `fecha` a tipo fecha.
2. Normalizar textos en `canal`.
3. Completar `metodo_pago` con `No informado`.
4. Completar variables numéricas con la mediana.

````python
# Crear una copia limpia
df_limpio = df.copy()
````

````python
# Converti fecha
df_limpio["fecha"] = pd.to_datetime(df_limpio["fecha"])
````

````python
# Normalizar textos en canal.
lista_caracteres = ("pais", "region", "ciudad", "canal", "categoria", "metodo_pago", "tipo_cliente", "devolucion")
for col in lista_caracteres:
  df_limpio[col] = df_limpio[col].astype("string").str.strip().str.title()
df_limpio.head()
````

````python
# Completar metodo_pago con No informado.
df_limpio["metodo_pago"] = df_limpio["metodo_pago"].fillna("No informado")
#print(df_limpio["metodo_pago"].isna().sum())
````

````python
# Completar variables numéricas con la mediana.
#mediana_diasentrega = df_limpio["dias_entrega"].median()
#mediana_edadcliente = df_limpio["edad_cliente"].median()
lista_cambios = ("dias_entrega", "edad_cliente", "satisfaccion")
for col in lista_cambios:
  mediana_col = df_limpio[col].median()
  df_limpio[col] = df_limpio[col].fillna(mediana_col)

````

````python
# Validación de resultados
nulos = pd.DataFrame({
    "nulos":      df_limpio.isna().sum(),
    "porcentaje": df_limpio.isna().mean() * 100
}).sort_values("nulos", ascending = False)
print(nulos)
#print(df_limpio["metodo_pago"].isna().sum())
````

Primero recorreremos columnas numéricas con `for`.

````python
columnas = df.select_dtypes(include="number").columns
for col in columnas:
  media = df_limpio[col].mean()
  mediana = df_limpio[col].median()
  min = df_limpio[col].min()
  max = df_limpio[col].max()
  print("valores de", col, ":", media, mediana, min, max)
````
